¿Qué representa una fila en cada archivo?
- En Afirmaciones, una fila es una afirmacion del dueño, con la cifra que declara y su unidad

```python
#Consulta 3
consulta_3 = pd.merge(compras, proveedores, on =  "proveedor_id", how = "inner")  #uno las dos tablas en la coincidencia de datos de proveedor_id, uso inner para quedarme solamente con los datos que se emparejan, que es lo que provoca el Join en SQL
consulta_3 = consulta_3[consulta_3["fecha_arribo"] >= "2025-01-01"] #limito los datos de fecha_arribo a las fechas superiores a 2025-01-01
consulta_3 = consulta_3 [["compra_id", "nombre", "pais", "fecha_arribo", "fecha_pedido"]] #selecciono las columnas que me quiero quedar
consulta_3 = consulta_3.rename (columns= {"nombre": "proveedor"}) #le cambio el nombre a la columna nombre por proveedor
consulta_3 = consulta_3.sort_values("fecha_arribo") #los ordeno por su fecha de arribo
consulta_3
```

```python
#consulta 5
consulta_5 = pd.merge(clientes, ventas, on = "cliente_id", how = "inner") #uno las tablas Clientes y ventas con la coincidencia en cliente_id
consulta_5 = consulta_5[consulta_5["estado"] == "confirmada"] #dejo en estado solo las confirmadas
consulta_5["facturacion"] = (consulta_5["cantidad"]*consulta_5["precio_unitario_ars"]*(1-consulta_5["descuento_pct"]/100)) #agrego la columna facturacion con lo calculado
consulta_5 = (consulta_5.groupby("provincia").agg(clientes=("cliente_id","nunique"), facturacion=("facturacion", "sum")).sort_values("facturacion", ascending = False))  #agrupo por provincia, sumo los clientes unicos y la facturacion total
consulta_5

```

```python
#Comparacion con la tabla ventas
ventas_confirmadas = ventas[ventas["estado"]=="confirmada"]
total_ventas = (ventas_confirmadas["cantidad"]* ventas_confirmadas["precio_unitario_ars"]*(1-ventas_confirmadas["descuento_pct"]/100)).sum()
total_consulta_5 = consulta_5["facturacion"].sum()
print("total en la tabla ventas:", round(total_ventas,2) )
print("total en la consulta 5:", round(total_consulta_5,2) )
print ("diferencia: ", round(total_ventas-total_consulta_5,2) )
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
ventas_confirmadas = ventas_limpias[ventas_limpias["estado"]=="confirmada"].copy()
ventas_confirmadas["mes"] = ventas_confirmadas["fecha"].dt.strftime("%Y-%m")
ventas_confirmadas["importes"] = ventas_confirmadas["cantidad"]*ventas_confirmadas["precio_unitario_ars"]*(1-ventas_confirmadas["descuento_pct"]/100)
```

```python
meses = pd.period_range("2023-01","2025-12", freq="M").strftime("%Y-%m")
serie = (ventas_confirmadas.groupby("mes").agg(facturacion=("importes","sum"), unidades = ("cantidad","sum")).reindex (meses,fill_value = 0))
serie.index.name = "mes"
assert len(serie) == 36
assert np.isclose(serie["facturacion"].sum(), ventas_confirmadas["importes"].sum())

x=range(len(serie))
fig, ax = plt.subplots(2, 1, figsize=(10, 7), sharex=True)
ax[0].plot(x, serie["facturacion"] / 1e6)
ax[0].set_title("Facturación mensual (pesos corrientes)")
ax[0].set_ylabel("millones de $")



ax[1].plot(x, serie["unidades"], color="tab:orange")
ax[1].set_title("Unidades vendidas por mes")
ax[1].set_ylabel("unidades")
ax[1].set_xticks(list(x)[::1])
ax[1].set_xticklabels(serie.index[::1], rotation=45)
plt.tight_layout()
plt.show()
```
